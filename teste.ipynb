```python
dados = []
encontrado = False
gatilho = "Y Axis Multiplier:\t1e-005"
df = pd.DataFrame()

with open(
    "C:\\Users\\jnrcr\\OneDrive\\Lacem\\Software SFC\\A-2025-090450-F004\\E01943\\SFC_FID_A-2025-090450-F004_LSM_20251203_Rep 1.dat.asc", "r", encoding="utf-8", errors="replace"
) as f:
    for linha in f:
        if not encontrado:
            if gatilho in linha:
                encontrado = True
            continue
        dados.append(linha.rstrip("\n"))

df["Intensidade"] = dados

df["Intensidade"] = (
    pd.to_numeric(df["Intensidade"], errors="coerce") * 1e-5
)

df["Tempo"] = (
    np.arange(1, len(df) + 1) * 0.016667 / 10
)
```
